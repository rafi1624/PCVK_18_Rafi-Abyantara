```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab.patches import cv2_imshow
from skimage import io

# 1. Unduh atau siapkan citra sampel
!wget -q -O sample.jpg https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg

img_bgr = cv2.imread('sample.jpg')
img_skimage = io.imread('sample.jpg')

# Output Soal 1
print("=== SOAL 1 ===")
print("Piksel [50, 50] pada pembacaan cv2 (BGR):", img_bgr[50, 50])
cv2_imshow(img_bgr)
plt.imshow(img_bgr)
plt.title("plt.imshow() tanpa konversi (menerima BGR)")
plt.show()

# Output Soal 2
print("\n=== SOAL 2 ===")
canvas_int16 = np.zeros((300, 300, 3), dtype=np.int16)
canvas_int32 = np.zeros((300, 300, 3), dtype=np.int32)
print(f"Canvas int16 -> dtype: {canvas_int16.dtype}, min: {canvas_int16.min()}, max: {canvas_int16.max()}, nbytes: {canvas_int16.nbytes} bytes")
print(f"Canvas int32 -> dtype: {canvas_int32.dtype}, min: {canvas_int32.min()}, max: {canvas_int32.max()}, nbytes: {canvas_int32.nbytes} bytes")

# Output Soal 4
print("\n=== SOAL 4 ===")
img_converted = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
print("Piksel [50, 50] skimage.io.imread():", img_skimage[50, 50])
print("Piksel [50, 50] cv2.cvtColor(BGR2RGB):", img_converted[50, 50])
print("Apakah seluruh array identik?", np.array_equal(img_skimage, img_converted))

# Output Soal 5
print("\n=== SOAL 5 ===")
print("img.shape awal:", img_bgr.shape)
plt.figure(figsize=(15, 15))
plt.imshow(img_converted)
plt.show()
print("img.shape setelah figsize besar:", img_bgr.shape)
```
